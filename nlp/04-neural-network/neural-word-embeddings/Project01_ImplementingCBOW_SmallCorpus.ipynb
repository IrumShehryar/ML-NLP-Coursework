# Sliding window of size 2 for each word

```python
neighbors = 2
for index, word in enumerate(tokenized_sentence):
    start = max(0, index - neighbors)
    finish = min(len(tokenized_sentence), index + neighbors + 1)
    print(word, tokenized_sentence[start:finish])
```

# Filter starting and ending words of sentences that do not have 2 neighbors around them

```python
for index, word in enumerate(tokenized_sentence):
    if (index < neighbors) or (len(tokenized_sentence)-index < neighbors+1):
     print(index, word)
     continue
```

# Generate the training data for CBOW. CBOW predicts the middle word based on the context. We are creating a training data in the form of dictionary where keys are the middle words and values are the neighbors

```python
training_data = {}

for index, word in enumerate(tokenized_sentence):
    if (index < neighbors) or (len(tokenized_sentence)-index < neighbors+1):
        continue
    else:
        start = index-neighbors
        finish = index+neighbors

        neighbor_words = tokenized_sentence[start:finish+1]

        print(word, neighbor_words[:neighbors]+neighbor_words[neighbors+1:])

        training_data[word] = neighbor_words[:neighbors]+neighbor_words[neighbors+1:]
```

```python
training_data # key is the word and value is the context
```

# Create one hot vectors for each word

```python
vocab_arrays = {}

for word, column_index in vocab_dict.items():
    word_array = np.zeros((1, len(vocab_dict)))
    word_array[0, column_index] = 1
    vocab_arrays[word] = word_array
```

# Sum of the Context Vectors

```python
for index, train in enumerate(training_data.items()):

    word = train[0]
    y[index,:] = vocab_arrays[word]

    temp_array = np.zeros([1, word_size])

    for neighbour in train[1]:
        temp_array = temp_array+vocab_arrays[neighbour]

    X[index,:] = temp_array
```

# Average of context vectors

```python
for index, train in enumerate(training_data.items()):

    word = train[0]
    y[index,:] = vocab_arrays[word]

    temp_array = np.zeros([1, word_size])

    for neighbour in train[1]:
        temp_array = temp_array+vocab_arrays[neighbour]

    X[index,:] = temp_array/(neighbors*2)
```

# Create and compile Neural Network

```python
from keras.models import Sequential
from keras.layers import Dense
```

```python
# Building model architecture
model = Sequential()
# Adding the hidden layer
model.add(Dense(4, input_dim=word_size, activation='relu')) # The 4 here is our choice. it is hidden layer.
# Adding the output layer
model.add(Dense(word_size, input_dim=4, activation='softmax'))

# Compiling the network
model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

# Train the NN

```python
# Fitting our model, let's train for 1000
# epochs
model.fit(X, y, epochs=1000)
```
