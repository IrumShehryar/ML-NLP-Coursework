```python
glove_model = {}
embedding = []
idx2word = []
with open('glove.6B.50d.txt', encoding='utf-8') as f:

  for line in f:
    values = line.split() # each line in file, we have space separated list of items
    word = values[0] # first item is a word
    vec = np.asarray(values[1:], dtype='float32') # After first item all the items are the components of word vector.
    # Now we have three kind of starage
    glove_model[word] = vec # word to vec storage where key is the word and value is the vector
    embedding.append(vec) # This is big embedding matrix, we can store as a list of numpy array so that we can casted to
                          # big numpy array later
    idx2word.append(word) # index to word mapping
print('%s word vectors are found.' % len(glove_model))
embedding = np.array(embedding)
V, D = embedding.shape
print(" ")
print("Embedding Shape :  ", V, D)
```

# Define a function for calculating analogies

```python
def analogies(word_1, word_2, word_3): # The input to the function are three words and we have to find the fourth word
  for word in (word_1, word_2, word_3):
    king = glove_model[word_1] # get the word vector for queen
  man = glove_model[word_2] # get the word vector for man
  woman = glove_model[word_3] # get the word vector for woman
  vec = king - man + woman # using king, man and woman, find vec

  # Pairwise_distance is used to find the pairwise distance between two vectors.
  # In our case the first vector is vec and second is all the word vectors in our embedding list.
  # The result of this multiplication is 1 x V.

  distances = pairwise_distances(vec.reshape(1, D), embedding, metric = 'cosine').reshape(V)
  indexes = distances.argsort()[:4] # Indices of the minimum distances
  for idx in indexes:
    WORD = idx2word[idx] # index to word dictionary to map back to the word
    if WORD not in (word_1, word_2, word_3):
      BEST_WORD = WORD
      break

  print(word_1, "-", word_2, "=", BEST_WORD, "-", word_3)
```
