# Find Analogies

### We can find the analogy of the word using the following equation
<br>
$ v^{king} + v^{woman} - v^{man}   \approx v^{queen} $


### Above equation states that if we add vectors of `king` and `woman` and subtract the vector of `man`, we will obtain a vector representation of the `queen'

```python
queens_vector = word2vec_model.most_similar(
    positive=['king', 'woman'],
    negative=['man']
)
```

```python
queens_vector
```

# Write a function to find the analogies

###  `word_1` - `word_2` = `?` - `word_3`
### e.g. `king` - `man` = `?` - `woman`
###      `?` = +`king` + `woman` -`man`.
### `king` and `woman` are `positive` i.e `word_1` and `word_3`.

```python
def analogies(word_1, word_2, word_3):
  pred = word2vec_model.most_similar(positive=[word_1, word_3], negative=[word_2])
  #print(pred)
  # pred contains multiple words with the scores. Thus to get the top words we need pred[0][0]
  print("%s - %s = %s - %s" % (word_1, word_2, pred[0][0], word_3))
```

```python
analogies('king', 'man', 'woman')
```

```python
analogies('france', 'paris', 'rome')
```
