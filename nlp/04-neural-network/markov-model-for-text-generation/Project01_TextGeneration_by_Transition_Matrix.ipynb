# Import Libraries

```python
import random
import numpy as np
import pandas as pd
```

# Create the text

```python
text = " The quick brown fox jumps over the lazy dog and the quick white cat jumps over the lazy rabbit "
```

# Adding start and end tokens to the text

```python
text = "<start>" + text  +  "<end>"
```

```python
text
```

# Tokenize the text

```python
tokens = text.lower().split()
```

```python
tokens
```

# Build the vocab

```python
vocab = list(set(tokens))
```

```python
vocab
```

```python
vocab.sort()
```

```python
vocab
```

# Count how many times the word occurs in vocab

```python
word_occurrences = {}

for token in vocab:
    word_occurrences[token] = tokens.count(token)

word_occurrences
```

# Build a transition Matrix

```python
transition_matrix = pd.DataFrame(
    np.zeros([len(vocab), len(vocab)]),
    columns=vocab,
    index=vocab
)
```

```python
transition_matrix
```

# Fill the Transition Matrix

```python
for index, token in enumerate(tokens):
    try:
        transition_matrix.loc[token, tokens[index+1]] += 1
        print(token, tokens[index+1])
        print(transition_matrix.loc[token, tokens[index+1]])
        print(" ")
    except IndexError:
        print('Reached end of the text!')
```

```python
transition_matrix
```

# Convert Transition Matrix into Probabilities

```python
for word in vocab:
    transition_matrix.loc[word,:] /= word_occurrences[word]
```

```python
transition_matrix
```

# Generate Text

```python
def get_next_token(token, transition_matrix):
    return transition_matrix.loc[token].sort_values().tail(1).index[0]
```

```python
sentence = ''
token = 'lazy'  #change starting token
```

```python
i = 0
while i < 20:
    sentence += token + ' '

    pred_token = get_next_token(token, transition_matrix)
    token = pred_token

    i += 1
```

```python
sentence
```

# Limitations
