# Import Libraries

```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

# Create text

```python
text = ['Hello my name is Zeeshan, this is Natural language processing class']
```

# Instantiate Count vectorizer and fit to the data

```python
vectorizer = CountVectorizer()
vector_matrix = vectorizer.fit_transform(text)
```

```python
vector_matrix.shape
```

```python
vector_matrix
```

```python
count_array = vector_matrix.toarray()
```

```python
count_array
```

```python
df = pd.DataFrame(data=count_array, index = ['Doc_0'], columns = vectorizer.get_feature_names_out())
```

```python
print(df) # Bag of Word
```

```python
print(vectorizer.vocabulary_) # Word2idx Mapping. The numbers do not represent the count but the position of the words in the matrix
```

```python
print(vector_matrix) # In (0, 1)  "0" is the document index and "1" is the position index of the word in the document/
                      #(0, 1) is repressenting the word hello.
```

# Another example

```python
text1 = ['I like vegetables',
        'I hate Eggs',
         'I like vegetables and I like Eggs']
```

```python
#vectorizer1 = CountVectorizer() # All single characters are ignored by default
```

```python
vectorizer1 = CountVectorizer(tokenizer=lambda txt: txt.split()) # if want to recognize single character then add tokenizer
```

```python
vector_matrix1 = vectorizer1.fit_transform(text1)
```

```python
count_array1 = vector_matrix1.toarray()
```

```python
count_array1 # Observe the sparsity in the matrix. If we have more doc and thus more words, we observe more sparsity.
```

```python
df1 = pd.DataFrame(data=count_array1, index = ['Doc_0','Doc_1', 'Doc_2'], columns = vectorizer1.get_feature_names_out())
```

```python
print(df1)
```

```python
print(vectorizer1.vocabulary_)
```

```python
# The numbers do not represent the count of the words but the position of the words in the matrix
```
