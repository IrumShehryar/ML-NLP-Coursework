```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch
```

```python
# Load the tokenizer and model for the specified pre-trained model
model_name = "distilbert-base-cased-distilled-squad"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

# Define a function to perform question answering manually
def Q_A(context, question):
    inputs = tokenizer(question, context, return_tensors="pt", truncation=True)
    with torch.no_grad():
        outputs = model(**inputs)

    answer_start_index = outputs.start_logits.argmax()
    answer_end_index = outputs.end_logits.argmax() + 1  # +1 for exclusive slicing

    # Get the tokens and decode the answer
    tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])
    answer_tokens = tokens[answer_start_index:answer_end_index]
    answer = tokenizer.decode(tokenizer.convert_tokens_to_ids(answer_tokens), skip_special_tokens=True)

    return {"answer": answer}
```

```python
context = "Today I teach transformers."      # This is answer
question = "What did students learn today?"
```

```python
Q_A(context=context, question=question)
```
