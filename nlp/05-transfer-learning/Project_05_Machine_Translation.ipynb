```python
from transformers import MarianMTModel, MarianTokenizer

# Load the model and tokenizer directly
model_name = 'Helsinki-NLP/opus-mt-en-fr'
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# Define a function to perform translation
def translator(text, **kwargs):
    # Prepare the input text for translation
    tokenized_text = tokenizer(text, return_tensors="pt")
    # Generate the translated tokens
    translated_tokens = model.generate(**tokenized_text, **kwargs)
    # Decode the translated tokens back to text
    return tokenizer.decode(translated_tokens[0], skip_special_tokens=True)
```

### English to Finnish Translation

Now, let's set up a translator for English to Finnish using a similar model.

```python
from transformers import MarianMTModel, MarianTokenizer

# Load the English to Finnish model and tokenizer
model_name_fi = 'Helsinki-NLP/opus-mt-en-fi'
tokenizer_fi = MarianTokenizer.from_pretrained(model_name_fi)
model_fi = MarianMTModel.from_pretrained(model_name_fi)

# Define a function to perform English to Finnish translation
def finnish_translator(text, **kwargs):
    # Prepare the input text for translation
    tokenized_text = tokenizer_fi(text, return_tensors="pt")
    # Generate the translated tokens
    translated_tokens = model_fi.generate(**tokenized_text, **kwargs)
    # Decode the translated tokens back to text
    return tokenizer_fi.decode(translated_tokens[0], skip_special_tokens=True)
```

You can now use `finnish_translator()` to translate text from English to Finnish. Here are a few examples:

```python
print(finnish_translator("Hello, how are you?"))
print(finnish_translator("I want to learn Finnish."))
```

```python
print(finnish_translator("My favourite subject is Neural Network"))
```
